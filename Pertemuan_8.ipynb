```python
#Nama:Ivandra
#Nim:4232501043
#Kelas:TRPE Pagi B

#6A
temps = [28, 31, 30, 33, 29]    # suhu (°C) selama 5 hari

print("temps[0] =", temps[0])   # elemen pertama
print("temps[4] =", temps[4])   # elemen terakhir (panjang list - 1)
print("temps[-1] =", temps[-1]) # indeks negatif: dihitung dari belakang

# --- prediksi dulu, baru jalankan ---
print(temps[1])                 # elemen pada indeks 1
print(temps[-2])                # elemen kedua dari belakang
print(temps[len(temps) - 1])    # elemen terakhir lewat len(temps) - 1
print(temps[2] + temps[3])      # jumlah elemen indeks 2 dan 3

# --- bandingkan prediksi dengan hasil ---
prediksi = [31, 33, 29, 63]     # prediksi saya untuk empat baris terakhir
hasil = [temps[1], temps[-2], temps[len(temps) - 1], temps[2] + temps[3]]

print()
for i in range(len(prediksi)):
    status = "sesuai" if prediksi[i] == hasil[i] else "BERBEDA"
    print("Baris", i + 1, "| prediksi:", prediksi[i], "| hasil:", hasil[i], "|", status)
```

```python
#Nama:Ivandra
#Nim:4232501043
#Kelas:TRPE Pagi B

#6F
scores = [80, 65, 90, 75, 70]

total = 0                    # akumulator: dimulai dari nol
count_pass = 0               # penghitung: akumulator yang menambah 1
for score in scores:
    total = total + score    # tambahkan tiap elemen
    if score >= 75:
        count_pass = count_pass + 1
    print("running total:", total)

average = total / len(scores)
print("Total   :", total)
print("Average :", average)
print("Scores >= 75:", count_pass, "of", len(scores))

# --- bandingkan prediksi dengan hasil ---
prediksi_total = 380
prediksi_rata = 76.0
prediksi_lulus = 3

print()
print("Total   :", "sesuai" if prediksi_total == total else "BERBEDA")
print("Rata-rata:", "sesuai" if prediksi_rata == average else "BERBEDA")
print("Jumlah >= 75:", "sesuai" if prediksi_lulus == count_pass else "BERBEDA")
```

```python
#Nama:Ivandra
#Nim:4232501043
#Kelas:TRPE Pagi B

#6G
temps = [28, 31, 30, 33, 29, 27, 32]

# Perubahan harian: mulai dari indeks 1 karena memakai i - 1
for i in range(1, len(temps)):
    change = temps[i] - temps[i - 1]
    print(f"hari {i + 1}: {temps[i]} °C, perubahan {change:+d}")

# Rata-rata bergerak 5 hari: jendela = 5 nilai yang berakhir di indeks i
avg5 = []
for i in range(4, len(temps)):          # mulai dari indeks 4 (sebelumnya 2)
    window = temps[i - 4 : i + 1]       # 5 nilai (sebelumnya i - 2)
    avg5.append(round(sum(window) / 5, 1))   # dibagi 5 (sebelumnya 3)

print("Rata-rata 5 hari:", avg5)
print(len(temps), "nilai ->", len(avg5), "rata-rata")
```

```python
#Nama:Ivandra
#Nim:4232501043
#Kelas:TRPE Pagi B

#6H
# Data energi harian (MWh), 1–30 November 2026
energy = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,   # hari 1–10
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,   # hari 11–20
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062,   # hari 21–30
]
run_hours = [24] * 30
run_hours[16] = 18            # hari ke-17 hanya beroperasi 18 jam
DAY_NAMES = ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"]
KAPASITAS = 330               # MW

i = 16                        # hari ke-17 = indeks 16

# 1) Perubahan dibanding hari sebelumnya
perubahan = (energy[i] - energy[i - 1]) / energy[i - 1] * 100
print(f"Hari {i + 1}: {energy[i]} MWh, perubahan {perubahan:+.1f}% dibanding hari {i}")

# 2) Jam operasi dan hari dalam seminggu
print("Jam operasi :", run_hours[i], "jam")
print("Hari        :", DAY_NAMES[i % 7])

# 3) Daya rata-rata selama beroperasi
daya_rata = energy[i] / run_hours[i]
print(f"Daya rata-rata: {daya_rata:.1f} MW dari kapasitas {KAPASITAS} MW")

# 4) Energi maksimum jika 18 jam berjalan penuh
maks_18 = KAPASITAS * run_hours[i]
print("Maksimum 18 jam pada beban penuh:", maks_18, "MWh")
print("Selisih yang tidak dijelaskan jam berhenti:", maks_18 - energy[i], "MWh")
```

```python
#Nama:Ivandra
#Nim:4232501043
#Kelas:TRPE Pagi B

#6I
temps = [29, 31, 28, 33, 30, 34, 27]                     # 7 suhu harian (°C)
DAY_NAMES = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]   # nama hari, indeks 0 = Senin

print("temps        :", temps)

# (1) Slicing
print("3 hari awal  :", temps[:3])      # indeks 0, 1, 2
print("2 hari akhir :", temps[-2:])     # dua elemen terakhir

# (2) append lalu pop
temps.append(32)                        # tambah satu hari di akhir
print("setelah append:", temps)
removed = temps.pop()                   # keluarkan elemen terakhir
print("di-pop        :", removed)
print("setelah pop   :", temps)

# (3) sorted: list baru yang terurut, temps tidak berubah
print("terurut       :", sorted(temps))
print("temps asli    :", temps)

# (4) rata-rata, suhu tertinggi, dan harinya
average = sum(temps) / len(temps)
highest = max(temps)
day_index = temps.index(highest)        # posisi suhu tertinggi = indeks hari
print("rata-rata     :", round(average, 2))
print("tertinggi     :", highest, "pada hari", DAY_NAMES[day_index], "(indeks", day_index, ")")

# (5) suhu di atas rata-rata (filter ke list baru)
above = []
for t in temps:
    if t > average:
        above.append(t)
print("di atas rata-rata:", above)
```

```python
#Nama:Ivandra
#Nim:4232501043
#Kelas:TRPE Pagi B

#WARM-UP
scores = [80, 65, 90, 75, 70, 85]

# (1) Skor pertama dan terakhir (indeks negatif untuk yang terakhir)
print("Skor pertama :", scores[0])
print("Skor terakhir:", scores[-1])

# (2) Rata-rata
rata_rata = sum(scores) / len(scores)
print("Rata-rata    :", rata_rata)

# (3) Banyak skor di atas rata-rata
jumlah_atas = 0
for s in scores:
    if s > rata_rata:
        jumlah_atas = jumlah_atas + 1
print("Skor di atas rata-rata:", jumlah_atas, "dari", len(scores))

# (4) Posisi skor terendah sebagai nomor siswa (siswa pertama = 1)
pos_min = 0
for i in range(len(scores)):
    if scores[i] < scores[pos_min]:
        pos_min = i
print("Skor terendah:", scores[pos_min], "milik siswa nomor", pos_min + 1)

# (5) Daftar baru dengan setiap skor ditambah 5
scores_baru = []
for s in scores:
    scores_baru.append(s + 5)
print("Skor + 5     :", scores_baru)
```

```python
#Nama:Ivandra
#Nim:4232501043
#Kelas:TRPE Pagi B

#LAB EXERCISE
energy = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,   # hari 1–10
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,   # hari 11–20
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062,   # hari 21–30
]

# 1) Total dan rata-rata energi
total = sum(energy)
rata_rata = total / len(energy)
print(f"Total energi bulan ini : {total:,} MWh")
print(f"Rata-rata energi       : {rata_rata:,.1f} MWh/hari")

# 2) Tiga hari tertinggi (cari posisi nilai terbesar, tandai, ulangi)
print("\nTiga hari tertinggi:")
sisa = energy.copy()                 # salinan, supaya energy asli tidak rusak
for k in range(3):
    pos = 0
    for i in range(len(sisa)):
        if sisa[i] > sisa[pos]:
            pos = i
    print(f"  {k + 1}. hari {pos + 1}: {energy[pos]:,} MWh")
    sisa[pos] = -1                   # tandai: sudah diambil

# 3) Tiga hari terendah
print("\nTiga hari terendah:")
sisa = energy.copy()
for k in range(3):
    pos = 0
    for i in range(len(sisa)):
        if sisa[i] < sisa[pos]:
            pos = i
    print(f"  {k + 1}. hari {pos + 1}: {energy[pos]:,} MWh")
    sisa[pos] = 10**9                # tandai: sudah diambil

# 4) Jumlah hari di bawah target yang dimasukkan pengguna
target = float(input("\nMasukkan target energi harian (MWh): "))
di_bawah = 0
for e in energy:
    if e < target:
        di_bawah = di_bawah + 1
print(f"Hari di bawah target {target:,.0f} MWh: {di_bawah} dari {len(energy)} hari")

# 5) Hari ketika rata-rata bergerak 7 hari < 97% dari rata-rata bulanan
batas = 0.97 * rata_rata
print(f"\nBatas 97% rata-rata bulanan: {batas:,.1f} MWh")
ada = False
for i in range(6, len(energy)):
    jendela = energy[i - 6 : i + 1]
    rata_7 = sum(jendela) / 7
    if rata_7 < batas:
        ada = True
        print(f"  Hari {i + 1} (hari {i - 5}–{i + 1}): rata-rata 7 hari {rata_7:,.1f} MWh")
if not ada:
    print("  Tidak ada.")

# 6) Total energi per minggu (minggu 1–4) + hari-hari yang tersisa
print("\nTotal energi per minggu:")
for minggu in range(4):
    awal = minggu * 7
    akhir = awal + 7
    print(f"  Minggu {minggu + 1} (hari {awal + 1}–{akhir}): {sum(energy[awal:akhir]):,} MWh")
sisa_hari = energy[28:]
print(f"  Sisa (hari 29–{len(energy)}): {sum(sisa_hari):,} MWh")
```

```python
# NAMA:IVANDRA
# NIM :4232501043

# Soal 2 Energi Harian dan Capacity Factor PLTD

daya_kw = 450
jam = 8
energi_aktual = 7200
kapasitas_kw = 400

energi = daya_kw * jam
print(f"Energi harian = {energi} kWh")

capacity_factor = (energi_aktual / (kapasitas_kw * 24)) * 100
print(f"Capacity factor = {capacity_factor:}%")

# Penjelasan
# Kode ini ngitung dua hal dari sebuah pembangkit listrik: energi harian dan capacity factor. Buat energi harian, tinggal kali daya 450 kW dengan 8 jam kerja,
# hasilnya 3.600 kWh. Buat capacity factor, energi aktual 7.200 kWh dibagi energi maksimum kalau pembangkit jalan penuh 24 jam (400 kW × 24 = 9.600 kWh),
# hasilnya 0,75 atau 75%. Jadi pembangkitnya hanya kepake 75% dari kemampuan maksimalnya.
```

```python
# Soal 4 Data Tekanan Uap Turbin

tekanan = [12.4, 13.1, 12.8, 14.2, 13.5]

print("Tekanan pembacaan ke-2:", tekanan[1])
print("Dua pembacaan terakhir:", tekanan[-2:])

tekanan.append(13.9)
tekanan.append(12.6)
print("Jumlah data setelah penambahan:", len(tekanan))

# Penjelasan
# Kode ini bikin list `tekanan` isi 5 data, nampilin data ke-2 (13.1) dan dua data terakhir ([14.2, 13.5]),
# terus nambahin 2 data baru lewat `append()` sampai totalnya jadi 7.
```

```python
# Soal 6 Status Operasi Boiler

suhu  = float(input("Masukkan suhu (C): "))
tekanan  = float(input("Masukkan tekanan (bar): "))

if suhu >= 550 or tekanan >= 14:
    print("STATUS: TRIP")
elif suhu > 520 and tekanan > 12:
    print("STATUS: WARNING")
else:
    print("STATUS: NORMAL")


# Penjelasan
# Kode ini minta input suhu (°C) dan tekanan (bar), lalu nentuin statusnya. Kalau suhu ≥ 550 **atau** tekanan ≥ 14,
# statusnya TRIP. Kalau nggak, tapi suhu > 520 **dan** tekanan > 12, statusnya WARNING. Selain itu, statusnya NORMAL.
```

```python
# Soal 8 Pengosongan Tangki Bahan Bakar

volume = 100

while volume > 0:
    volume -= 20
    print(f"Volume: {volume} liter")
    if volume == 20:
        print("WARNING: Hampir kosong!")
        print("TANGKI KOSONG - Pengosongan berhenti.")

# Penjelasan
# Kode ini ngurangin `volume` dari 100 liter, 20 tiap putaran, sampai 0 (80, 60, 40, 20, 0).
# Pas volume 20, muncul warning plus pesan "TANGKI KOSONG", padahal belum kosong, jadi pesannya salah tempat.
```

```python
# Soal 10 Pembacaan Suhu Bearing Dengan Sensor Rusak

data = ["62.5", "58.0", "81.3", "N/A", "66.0"]
jumlah_valid = 0

for i in range(len(data)):
    n = i + 1
    try:
        nilai = float(data[i])
    except ValueError:
        print(f"Titik {n}: data tidak valid - pemeriksaan dihentikan")
        break
    jumlah_valid += 1
    if nilai > 70.0:
        print(f"Warning: Titik {n} suhu bearing tinggi!")
print(f"Jumlah data valid: {jumlah_valid}")

# Penjelasan
# Kode ini ngecek data suhu bearing satu per satu. Titik 3 (81.3) lebih dari 70 jadi kena warning, terus di titik 4 isinya "N/A" jadi error dan loop berhenti, sehingga titik 5 nggak dicek.
# Hasil akhirnya, jumlah data valid: 3.
```

```python
# SOAL 1:Belanja Buku
jumlah_buku = int(input("Jumlah buku           : "))
harga_buku  = int(input("Harga per buku (Rp)   : "))
diskon_persen = int(input("Diskon (%)           : "))
uang        = int(input("Uang Budi (Rp)        : "))
jumlah_orang = int(input("Jumlah orang patungan : "))

total     = jumlah_buku * harga_buku
diskon    = total * diskon_persen // 100
bayar     = total - diskon
kembalian = uang - bayar
per_orang = bayar // jumlah_orang
sisa      = bayar % jumlah_orang
cukup     = uang >= bayar

print("Total       :", total)
print("Diskon      :", diskon)
print("Bayar       :", bayar)
print("Kembalian   :", kembalian)
print("Per orang   :", per_orang)
print("Sisa        :", sisa)
print("Uang cukup? :", cukup)
```

```python
# SOAL 2:PLTS Atap Rumah

panel      = int(input("Jumlah panel surya        : "))
daya_panel = int(input("Daya per panel (Wp)       : "))
jam        = int(input("Penyinaran efektif (jam)  : "))
efisiensi  = float(input("Efisiensi sistem (0-1)    : "))
kebutuhan  = float(input("Kebutuhan listrik (kWh)   : "))
tarif      = float(input("Tarif listrik (Rp/kWh)    : "))

hari       = 30
harian    = panel * daya_panel * jam * efisiensi / 1000
bulanan   = harian * hari
nilai     = bulanan * tarif
kurang    = kebutuhan - harian
mencukupi = harian >= kebutuhan

print()
print("Energi harian (kWh) :", round(harian, 2))
print("Energi bulanan (kWh):", round(bulanan, 2))
print("Nilai listrik (Rp)  :", round(nilai, 2))
print("Kekurangan (kWh)    :", round(kurang, 2))
print("Mencukupi?          :", mencukupi)
```

```python
# SOAL 3:Generator Diesel dan Rugi Kabel
daya      = int(input("Daya generator (kW)        : "))
jam       = int(input("Lama operasi (jam/hari)    : "))
konsumsi  = float(input("Konsumsi (liter/kWh)       : "))
harga     = float(input("Harga solar (Rp/liter)     : "))
tangki    = float(input("Kapasitas tangki (liter)   : "))
arus      = int(input("Arus kabel (A)             : "))
hambatan  = float(input("Hambatan kabel (ohm)       : "))

energi    = daya * jam
solar     = energi * konsumsi
biaya     = solar * harga
hari_cukup = tangki // solar
sisa_solar = tangki % solar
rugi      = arus ** 2 * hambatan
tinggi    = rugi > 150 and arus >= 20

print()
print("Energi (kWh)     :", energi)
print("Solar (liter)    :", solar)
print("Biaya (Rp)       :", biaya)
print("Tangki cukup     :", hari_cukup, "hari")
print("Sisa solar (L)   :", sisa_solar)
print("Rugi kabel (W)   :", rugi)
print("Rugi tinggi?     :", tinggi)
```

```python
# Nama: [Ivandra]
# NIM: [4232501043]
# Tanggal: [27/09/20]
# Tujuan: Menampilkan hasil operasi +, -, *, /, //, dan % dari dua bilangan bulat

# Jawaban 1

a = int(input("Masukkan bilangan pertama: "))
b = int(input("Masukkan bilangan kedua: "))

print(f"\nBilangan pertama : {a}")
print(f"Bilangan kedua   : {b}\n")

print(f"Penjumlahan (+)       : {a + b}")
print(f"Pengurangan (-)       : {a - b}")
print(f"Perkalian (*)         : {a * b}")
print(f"Pembagian (/)         : {a / b}")
print(f"Pembagian bulat (//)  : {a // b}")
print(f"Modulus (%)           : {a % b}")
```

```python
# Nama: [Ivandra]
# NIM: [4232501043]
# Tanggal: [27/09/2026]
# Tujuan: Mengecek kategori kelulusan berdasarkan nilai ujian menggunakan operator relasional

# Jawaban 2

nilai = float(input("Masukkan nilai ujian (0-100): "))

lulus = nilai >= 60

print(f"\nNilai ujian yang dimasukkan : {nilai}")
print(f"Nilai ini termasuk kategori lulus (>=60): {lulus}")
```

```python
# Nama: [Ivandra]
# NIM: [4232501043]
# Tanggal: [27/09/2026]
# Tujuan: Menentukan kelayakan mengemudi sendiri menggunakan operator logika (and)

# Jawaban 3

umur = int(input("Masukkan umur: "))
punya_sim = input("Punya SIM? (True/False): ") == "True"

# Tanda kurung digunakan untuk memperjelas urutan evaluasi (latihan operator precedence)
boleh_mengemudi = (umur >= 17) and (punya_sim == True)

print(f"\nUmur           : {umur}")
print(f"Punya SIM      : {punya_sim}")
print(f"Boleh mengemudi sendiri: {boleh_mengemudi}")
```

```python
# Nama: [Ivandra]
# NIM: [4232501043]
# Tanggal: [27/09/2026]
# Tujuan: Menghitung efisiensi termal unit pembangkit

# Jawaban 4

energi_input = float(input("Masukkan energi input (MJ): "))
energi_output = float(input("Masukkan energi output (MJ): "))

# Tanda kurung memastikan pembagian dilakukan sebelum dikali 100
efisiensi = (energi_output / energi_input) * 100

print(f"\nEnergi input  : {energi_input} MJ")
print(f"Energi output : {energi_output} MJ")
print(f"Efisiensi termal unit pembangkit: {efisiensi:.2f}%")
```

```python
# Nama: [Ivandra]
# NIM: [4232501043]
# Tanggal: [27/09/2026]
# Tujuan: Mengecek status kebutuhan maintenance unit pembangkit menggunakan operator relasional dan logika

# Jawaban 5

suhu_boiler = float(input("Masukkan suhu boiler (°C): "))
beban_operasi = float(input("Masukkan beban operasi (%): "))

perlu_maintenance = (suhu_boiler > 550) or (beban_operasi > 95)

print(f"\nSuhu boiler     : {suhu_boiler}°C")
print(f"Beban operasi   : {beban_operasi}%")
print(f"Unit pembangkit dengan suhu boiler {suhu_boiler}°C dan beban operasi {beban_operasi}% "
      f"berstatus perlu maintenance: {perlu_maintenance}")
```
